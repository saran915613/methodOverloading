```python

from collections import defaultdict

# 1. Initialize with 'set' as the default factory
skaters_by_country = defaultdict(set)

# 2. Add data directly (even if the key doesn't exist yet!)
# Use .add() for sets
skaters_by_country["Japan"].add("Yuzuru Hanyu")
skaters_by_country["Japan"].add("Shoma Uno")
skaters_by_country["USA"].add("Nathan Chen")

# Duplicates are automatically ignored by the set:
skaters_by_country["Japan"].add("Yuzuru Hanyu")

print(dict(skaters_by_country))
# Output: {'Japan': {'Yuzuru Hanyu', 'Shoma Uno'}, 'USA': {'Nathan Chen'}}
```

### Use defaultdict(list) to collect elements and cast them to fixed tuples when finished

```python
from collections import defaultdict

# 1. Initialize with 'list' so we can dynamically add items
skater_medals = defaultdict(list)

# 2. Append data dynamically
skater_medals["Hanyu"].append("Gold")
skater_medals["Hanyu"].append("Gold")
skater_medals["Chen"].append("Gold")

# 3. Convert to fixed tuples when finished
final_medals = {key: tuple(value) for key, value in skater_medals.items()}

print(final_medals)
# Output: {'Hanyu': ('Gold', 'Gold'), 'Chen': ('Gold',)}
```

### Zip parallel lists of keys and values together to populate a grouped defaultdict

```python
from collections import defaultdict

# Raw parallel datasets
countries = ["Japan", "USA", "Japan", "USA"]
skaters = ["Yuzuru Hanyu", "Nathan Chen", "Shoma Uno", "Ilia Malinin"]

# 1. Initialize the defaultdict
grouped_skaters = defaultdict(list)

# 2. Use zip() to loop through both lists simultaneously
for country, skater in zip(countries, skaters):
    grouped_skaters[country].append(skater)

print(dict(grouped_skaters))
# Output: {'Japan': ['Yuzuru Hanyu', 'Shoma Uno'], 'USA': ['Nathan Chen', 'Ilia Malinin']}
```

### Pre-process list elements with map() before zipping and grouping them into a dict

```python
from collections import defaultdict

# Raw uncleaned list
raw_skaters = ["yuzuru hanyu", "nathan chen", "shoma uno"]
countries = ["Japan", "USA", "Japan"]

# 1. Initialize the defaultdict
cleaned_groups = defaultdict(list)

# 2. Use map() with a text transformation function (e.g., .title())
# map() applies string.title to every single element automatically
cleaned_skaters = map(str.title, raw_skaters)

# 3. Zip the mapped result with the keys
for country, skater in zip(countries, cleaned_skaters):
    cleaned_groups[country].append(skater)

print(dict(cleaned_groups))
# Output: {'Japan': ['Yuzuru Hanyu', 'Shoma Uno'], 'USA': ['Nathan Chen']}
```

### Map a lambda function over a list of strings to construct a dictionary of name lengths

```python
skaters = ["Yuzuru Hanyu", "Nathan Chen", "Shoma Uno"]

# Iterate over the list. Lambda turns each name into a (name, length) pair
name_lengths = dict(map(lambda name: (name, len(name)), skaters))

print(name_lengths)
# Output: {'Yuzuru Hanyu': 12, 'Nathan Chen': 11, 'Shoma Uno': 9}
```

### Use map() with lambda to append structured pairs directly into a defaultdict

```python
If your data is already structured as pairs, you can use map() with a lambda function to feed data directly into your defaultdict without an explicit for loop, though a standard loop is often preferred for readability:
python
from collections import defaultdict

data_pairs = [("Japan", "Yuzuru Hanyu"), ("USA", "Nathan Chen"), ("Japan", "Shoma Uno")]
skater_map = defaultdict(list)

# map passes each pair into the lambda, which appends it to the list
list(map(lambda pair: skater_map[pair[0]].append(pair[1]), data_pairs))

print(dict(skater_map))
# Output: {'Japan': ['Yuzuru Hanyu', 'Shoma Uno'], 'USA': ['Nathan Chen']}
```

### Sort a dictionary by its values in ascending order using sorted() with a lambda key

```python
skater_ranks = {"Shoma Uno": 3, "Nathan Chen": 1, "Yuzuru Hanyu": 2}

# sorted() iterates through the items.
# The lambda x: x[1] tells it to sort based on index 1 (the values/ranks)
sorted_by_rank = dict(sorted(skater_ranks.items(), key=lambda item: item[1]))

print(sorted_by_rank)
# Output: {'Nathan Chen': 1, 'Yuzuru Hanyu': 2, 'Shoma Uno': 3}
```

### Map a custom function over a list to modify and return a new lazy generator

```python
########  MAP

# Our machine (function) that doubles a number
def double_me(number):
    return number * 2

# Our raw data list
my_numbers = [1, 2, 3, 4]

# Pass the machine and the numbers to map()
result = map(double_me, my_numbers)

# map() returns a generator object, so we wrap it in list() to see the answer
print(list(result))
# Output: [2, 4, 6, 8]
```

**lambda inside a dictionary comprehension to iterate and add bonus points**

```python
# Raw scores dataset
raw_scores = {"Hanyu": 110.5, "Chen": 111.0, "Uno": 105.2}

# Use lambda inside a dictionary comprehension to iterate and add bonus points
# pair[0] is the Key, pair[1] is the Value
adjusted_scores = dict(map(lambda pair: (pair[0], pair[1] + 2.0), raw_scores.items()))

print(adjusted_scores)
# Output: {'Hanyu': 112.5, 'Chen': 113.0, 'Uno': 107.2}
```

```python
def compute_ranking(scores):
    # Step 1: Infer the number of judges dynamically from the first skater's score list length
    # Each skater has 2 lists (Tech, Pres). The length of list[0] is the number of judges.
    sample_skater = list(scores.keys())[0]
    num_judges = len(scores[sample_skater][0])

    # Step 2: Initialize a dictionary to store unsorted data for each judge
    ranking_by_judge = {}
    for i in range(num_judges):
        ranking_by_judge[f"J{i}"] = []

    # Step 3: Loop through the skaters and calculate totals for each judge
    for skater, score_lists in scores.items():
        tech_scores = score_lists[0]
        pres_scores = score_lists[1]

        for i in range(num_judges):
            # Sum the tech and presentation score for Judge i, rounded to 2 decimal places
            total_score = round(tech_scores[i] + pres_scores[i], 2)
            # Append the tuple (Skater Name, Total Score) to Judge i's list
            ranking_by_judge[f"J{i}"].append((skater, total_score))

    # Step 4: Sort each judge's list using the tie-breaker logic
    final_rankings = {}
    for judge, skater_tuples in ranking_by_judge.items():
        # Lambda rule: Sort by negative score (forces highest first),
        # then by name (forces alphabetical A-Z first)
        sorted_list = sorted(skater_tuples, key=lambda x: (-x[1], x[0]))
        final_rankings[judge] = sorted_list

    return final_rankings


# --- Exercise Demo ---
scores = {
    'Plushenko': [[5.7, 5.8, 6.0], [5.2, 5.3, 6.0]],
    'Goebel': [[5.6, 5.0, 5.6], [5.2, 5.4, 5.4]],
    'Honda': [[5.7, 5.2, 5.6], [5.1, 5.4, 5.4]],
    'Yagudin': [[5.9, 5.9, 5.9], [5.3, 5.4, 5.5]]
}

ranking_by_judge = compute_ranking(scores)
import pprint
pprint.pprint(ranking_by_judge)
```
